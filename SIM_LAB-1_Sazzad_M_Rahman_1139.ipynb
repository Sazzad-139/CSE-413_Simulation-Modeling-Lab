### What is Simulation and Modeling?

**Simulation and modeling** are essential techniques used to understand, predict, and improve the behavior of complex systems (mathematical models). In **modeling**, you create a simplified version of a system—like a blueprint or map—that represents its key components. In **simulation**, you run experiments with that model to observe how it behaves over time under various conditions.

For example, **in a traffic management system**, you might model how cars flow through an intersection, and then simulate what happens during rush hour or in the case of an accident. Instead of testing the system in real life (which could be dangerous or costly), you can test it using **Python** simulations.

# Task 1: Scalar and Array Built-in Mathematical Functions

```python
# Required scalar value and a custom array containing at least 6 elements
scalar = 3.7
arr = np.array([0.5, 1.2, 2.8, 3.6, 4.5, 6.0])

print("Scalar =", scalar)
print("Array =", arr)

# A helper function prints the same mathematical operations for any input
def show_math_operations(name, values):
    values = np.asarray(values)

    # Clip only for inverse sine and inverse cosine because their valid domain is [-1, 1]
    clipped_values = np.clip(values, -1, 1)

    print(f"\n--- Mathematical operations on {name} ---")
    print("sin:", np.sin(values))
    print("cos:", np.cos(values))
    print("tan:", np.tan(values))
    print("arcsin (clipped):", np.arcsin(clipped_values))
    print("arccos (clipped):", np.arccos(clipped_values))
    print("arctan:", np.arctan(values))
    print("exp:", np.exp(values))
    print("log (natural):", np.log(values))
    print("absolute:", np.abs(values))
    print("square root:", np.sqrt(values))
    print("remainder when divided by 2:", np.remainder(values, 2))
    print("round:", np.round(values))
    print("floor:", np.floor(values))
    print("ceil:", np.ceil(values))

# Display all required function outputs for both the scalar and array
show_math_operations("scalar", scalar)
show_math_operations("array", arr)
```

## Do it For 2X2 matrix

```python
# Creating a 2x2 matrix
matrix = np.array([[1, 2], [3, 4]])

# Trigonometric functions
sin_values = np.sin(matrix)
cos_values = np.cos(matrix)
tan_values = np.tan(matrix)
asin_values = np.arcsin(np.clip(matrix / np.max(matrix), -1, 1))  # Clipping for valid range [-1, 1]
acos_values = np.arccos(np.clip(matrix / np.max(matrix), -1, 1))
atan_values = np.arctan(matrix)

# Exponential and logarithm
exp_values = np.exp(matrix)
log_values = np.log(matrix)  # Natural logarithm (log base e)

# Absolute value and square root
abs_values = np.abs(matrix)
sqrt_values = np.sqrt(matrix)

# Remainder when divided by a scalar (e.g., 2)
rem_values = np.remainder(matrix, 2)

# Rounding functions
round_values = np.round(matrix)
floor_values = np.floor(matrix)
ceil_values = np.ceil(matrix)

# Display results
print("Matrix:\n", matrix)
print("\nsin:\n", sin_values)
print("\ncos:\n", cos_values)
print("\ntan:\n", tan_values)
print("\nasine:\n", asin_values)
print("\nacosine:\n", acos_values)
print("\natangent:\n", atan_values)
print("\nexp:\n", exp_values)
print("\nlog (natural):\n", log_values)
print("\nabs:\n", abs_values)
print("\nsqrt:\n", sqrt_values)
print("\nrem (remainder when divided by 2):\n", rem_values)
print("\nround:\n", round_values)
print("\nfloor:\n", floor_values)
print("\nceil:\n", ceil_values)
```

# Task 2: Apply the Same Operations to a 3×2 Matrix

```python
# Create a custom 3x2 matrix with positive floating-point values
matrix_3x2 = np.array([
    [0.5, 1.2],
    [2.7, 3.4],
    [4.8, 5.6]
])

# Clip the original matrix for arcsin and arccos to prevent domain errors
clipped_matrix = np.clip(matrix_3x2, -1, 1)

# Perform all required operations
sin_matrix = np.sin(matrix_3x2)
cos_matrix = np.cos(matrix_3x2)
tan_matrix = np.tan(matrix_3x2)
asin_matrix = np.arcsin(clipped_matrix)
acos_matrix = np.arccos(clipped_matrix)
atan_matrix = np.arctan(matrix_3x2)

exp_matrix = np.exp(matrix_3x2)
log_matrix = np.log(matrix_3x2)

abs_matrix = np.abs(matrix_3x2)
sqrt_matrix = np.sqrt(matrix_3x2)

remainder_matrix = np.remainder(matrix_3x2, 2)

round_matrix = np.round(matrix_3x2)
floor_matrix = np.floor(matrix_3x2)
ceil_matrix = np.ceil(matrix_3x2)

# Display the matrix and all results
print("Original 3x2 matrix:\n", matrix_3x2)
print("\nSine:\n", sin_matrix)
print("\nCosine:\n", cos_matrix)
print("\nTangent:\n", tan_matrix)
print("\nArcsine (after clipping to [-1, 1]):\n", asin_matrix)
print("\nArccosine (after clipping to [-1, 1]):\n", acos_matrix)
print("\nArctangent:\n", atan_matrix)
print("\nExponential:\n", exp_matrix)
print("\nNatural logarithm:\n", log_matrix)
print("\nAbsolute value:\n", abs_matrix)
print("\nSquare root:\n", sqrt_matrix)
print("\nRemainder when divided by 2:\n", remainder_matrix)
print("\nRound:\n", round_matrix)
print("\nFloor:\n", floor_matrix)
print("\nCeil:\n", ceil_matrix)
```

# Vector Functions

```python
# Sample vector (1D array)
vec = np.array([3, 1, 4, 1, 5, 9, 2, 6])

# 1. Maximum value and index of max element
max_value = np.max(vec)
max_index = np.argmax(vec)  # Index of max element

# 2. Minimum value and index of min element
min_value = np.min(vec)
min_index = np.argmin(vec)  # Index of min element

# 3. Length of the vector
vec_length = len(vec)  # Equivalent to MATLAB's length()

# 4. Sorting in ascending order
sorted_vec = np.sort(vec)



# 5. Sum of elements
sum_values = np.sum(vec)

# 6. Product of elements
prod_values = np.prod(vec)

# 7. Median value
median_value = np.median(vec)

# 8. Mean value
mean_value = np.mean(vec)

# 9. Standard deviation
std_dev = np.std(vec)

# Display results
print("Max value:", max_value, "at index", max_index)
print("Min value:", min_value, "at index", min_index)
print("Length of vector:", vec_length)
print("Sorted vector:", sorted_vec)
print("Sum of elements:", sum_values)
print("Product of elements:", prod_values)
print("Median value:", median_value)
print("Mean value:", mean_value)
print("Standard deviation:", std_dev)
```

# Task 3: Vector Functions on a 2×3 Floating-point Matrix

```python
# Create the required 2x3 matrix using floating-point values
matrix_2x3 = np.array([
    [2.1, 4.7, 1.5],
    [6.3, 3.8, 5.2]
])

# Maximum and minimum values and their 2D indices
max_value = np.max(matrix_2x3)
max_index = np.unravel_index(np.argmax(matrix_2x3), matrix_2x3.shape)

min_value = np.min(matrix_2x3)
min_index = np.unravel_index(np.argmin(matrix_2x3), matrix_2x3.shape)

# For a matrix, size gives the total number of elements.
# The Euclidean/Frobenius norm is also shown as a mathematical length.
total_elements = matrix_2x3.size
matrix_length = np.linalg.norm(matrix_2x3)

# Sort each row in descending order
row_wise_descending = np.sort(matrix_2x3, axis=1)[:, ::-1]

# Flatten, sort the entire matrix in descending order, and keep it as a 1D array
entire_matrix_descending = np.sort(matrix_2x3.flatten())[::-1]

# Statistical and aggregate operations
sum_value = np.sum(matrix_2x3)
product_value = np.prod(matrix_2x3)
median_value = np.median(matrix_2x3)
mean_value = np.mean(matrix_2x3)
std_value = np.std(matrix_2x3)

# Display all required results
print("Original 2x3 matrix:\n", matrix_2x3)
print("\nMaximum value:", max_value, "at index", max_index)
print("Minimum value:", min_value, "at index", min_index)
print("Total number of elements:", total_elements)
print("Matrix length (Euclidean/Frobenius norm):", matrix_length)
print("\nRow-wise descending sort:\n", row_wise_descending)
print("\nEntire matrix descending sort:\n", entire_matrix_descending)
print("\nSum:", sum_value)
print("Product:", product_value)
print("Median:", median_value)
print("Mean:", mean_value)
print("Standard deviation:", std_value)
```

# Bonus 1: Visualize Matrix Values with a Heatmap

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(413)

random_matrix = np.random.uniform(0.5, 10.0, size=(3, 2))

print("Random 3x2 matrix:\n", random_matrix)
print("\nMaximum:", np.max(random_matrix))
print("Minimum:", np.min(random_matrix))
print("Mean:", np.mean(random_matrix))
print("Standard deviation:", np.std(random_matrix))
print("Row-wise descending:\n", np.sort(random_matrix, axis=1)[:, ::-1])

# Visualize the random matrix
plt.figure(figsize=(6, 4))
sns.heatmap(
    random_matrix,
    annot=True,
    fmt=".2f",
    cmap="viridis",
    cbar=True,
    linewidths=0.5
)
plt.title("Heatmap of Randomly Generated Matrix")
plt.xlabel("Column")
plt.ylabel("Row")
plt.show()
```
